# Email Spam Detection Using NFA

An educational keyword-matching demonstration using a Non-Deterministic Finite Automaton (NFA). It scans the sender, subject, and body for configured spam-related words. This is not a production email filter.

```python
SPAM_KEYWORDS = (
    "free", "win", "offer", "prize", "claim", "urgent",
    "lottery", "cash", "click", "congratulations", "verify", "suspended"
)

class EmailSpamNFA:
    def __init__(self, keywords=SPAM_KEYWORDS):
        self.keywords = tuple(word.lower() for word in keywords)

    def scan(self, text):
        text = text.lower()
        active_states = set()
        matched = set()
        for character in text:
            next_active = set()
            # Nondeterministically start a keyword path at each character.
            for keyword in self.keywords:
                if keyword[0] == character:
                    if len(keyword) == 1:
                        matched.add(keyword)
                    else:
                        next_active.add((keyword, 1))
            # Advance all active keyword paths.
            for keyword, position in active_states:
                if keyword[position] == character:
                    if position + 1 == len(keyword):
                        matched.add(keyword)
                    else:
                        next_active.add((keyword, position + 1))
            active_states = next_active
        return sorted(matched)

    def classify(self, sender, subject, body):
        matches = self.scan(" ".join((sender, subject, body)))
        return ("SPAM / SUSPICIOUS" if matches else "NOT SPAM", matches)

detector = EmailSpamNFA()
print("NFA email spam detector is ready.")
```

## Try an email
Edit the three values below and run the cell.

```python
sender = "offers@example.com"
subject = "Congratulations! You have won a prize"
body = "Click now to claim your free gift."

classification, matched_words = detector.classify(sender, subject, body)
print("Classification:", classification)
print("Matched words:", ", ".join(matched_words) if matched_words else "None")
```
